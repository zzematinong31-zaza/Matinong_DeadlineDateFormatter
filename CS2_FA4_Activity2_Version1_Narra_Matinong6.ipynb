```python
# Asks for user's input
event_title = input("Enter event title: ")
year = input("Enter year: ")
month = int(input("Enter month (1-12):"))
day = input("Enter day: ")

# Used selction structure
if month == 1:
  month_name = "January"
elif month == 2:
  month_name = "February"
elif month == 3:
  month_name = "March"
elif month == 4:
  month_name = "April"
elif month == 5:
  month_name = "May"
elif month == 6:
  month_name = "June"
elif month == 7:
  month_name = "July"
elif month == 8:
  month_name = "August"
elif month == 9:
   month_name = "September"
elif month == 10:
  month_name = "October"
elif month == 11:
  month_name = "November"
elif month == 12:
  month_name = "December"

else:
  month_name = "Invalid Month"

# Output without date formatting

print(" ")
print("Event:", event_title.title())
print("Date:", month_name, day, ",", year)
print("Year:", year)
print("Month:", month_name)
print("Day:", day)
```
